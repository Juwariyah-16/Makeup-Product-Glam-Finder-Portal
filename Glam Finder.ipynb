# GLAM FINDER
### Makeup Product Finder Portal

Glam Finder is a Python-based makeup product finder
portal that helps users search, explore, and manage
different makeup products.
The portal allows users to find products according
to their category, brand, skin type, and budget.

**Developed using:** Python and JSON

```python
import json
import os
```

```python
products = [
    {
        "id": 1,
        "name": "Matte Foundation",
        "brand": "Maybelline",
        "category": "Foundation",
        "skin_type": "Oily",
        "price": 2500
    },
    {
        "id": 2,
        "name": "Fit Me Concealer",
        "brand": "Maybelline",
        "category": "Concealer",
        "skin_type": "All",
        "price": 1800
    },
    {
        "id": 3,
        "name": "Soft Blush",
        "brand": "Miss Rose",
        "category": "Blush",
        "skin_type": "All",
        "price": 1200
    },
    {
        "id": 4,
        "name": "Lip Gloss",
        "brand": "NYX",
        "category": "Lipstick",
        "skin_type": "All",
        "price": 2200
    },
    {
        "id": 5,
        "name": "Compact Powder",
        "brand": "Rivaj",
        "category": "Powder",
        "skin_type": "Oily",
        "price": 900
    }
]

print("Product database created successfully!")
```

```python
try:
    with open("products.json", "r") as file:
        products = json.load(file)

    print("Existing products loaded successfully!")

except FileNotFoundError:
    with open("products.json", "w") as file:
        json.dump(products, file, indent=4)

    print("New product database created!")

except json.JSONDecodeError:
    print("Database file contains invalid data!")

except PermissionError:
    print("Permission denied!")
```

```python
def view_products():
    print("\n===== ALL MAKEUP PRODUCTS =====")

    for product in products:
        print("----------------------------")
        print("Product ID:", product["id"])
        print("Name:", product["name"])
        print("Brand:", product["brand"])
        print("Category:", product["category"])
        print("Skin Type:", product["skin_type"])
        print("Price: Rs.", product["price"])

    print("----------------------------")
```

```python
view_products()
```

```python
def search_product():
    search_name = input("Enter product name: ")

    found = False

    for product in products:
        if search_name.lower() in product["name"].lower():
            print("\nProduct Found!")
            print("Name:", product["name"])
            print("Brand:", product["brand"])
            print("Category:", product["category"])
            print("Price: Rs.", product["price"])

            found = True

    if found == False:
        print("Product not found!")
```

```python
search_product()
```

```python
def filter_category():
    category = input("Enter makeup category: ")

    found = False

    for product in products:
        if product["category"].lower() == category.lower():
            print("\nProduct Found!")
            print("Name:", product["name"])
            print("Brand:", product["brand"])
            print("Price: Rs.", product["price"])

            found = True

    if found == False:
        print("No products available in this category.")
```

```python
filter_category()
```

```python
def filter_budget():
    try:
        budget = float(input("Enter your maximum budget (Rs.): "))

        found = False

        for product in products:
            if product["price"] <= budget:
                print("\nProduct Found!")
                print("Name:", product["name"])
                print("Brand:", product["brand"])
                print("Price: Rs.", product["price"])

                found = True

        if found == False:
            print("No products found within this budget.")

    except ValueError:
        print("Please enter a valid number!")
```

```python
filter_budget()
```

```python
def filter_skin_type():
    skin = input("Enter your skin type (Oily, Dry, Combination): ")

    found = False

    for product in products:
        if product["skin_type"].lower() == skin.lower() or product["skin_type"].lower() == "all":
            print("\nProduct Found!")
            print("Name:", product["name"])
            print("Brand:", product["brand"])
            print("Skin Type:", product["skin_type"])
            print("Price: Rs.", product["price"])

            found = True

    if found == False:
        print("No products found for this skin type.")
```

```python
filter_skin_type()
```

```python
def add_product():
    try:
        new_product = {}

        new_product["id"] = len(products) + 1
        new_product["name"] = input("Enter product name: ")
        new_product["brand"] = input("Enter brand: ")
        new_product["category"] = input("Enter category: ")
        new_product["skin_type"] = input("Enter skin type: ")
        new_product["price"] = float(input("Enter price: "))

        products.append(new_product)

        with open("products.json", "w") as file:
            json.dump(products, file, indent=4)

        print("\nProduct added successfully!")

    except ValueError:
        print("Please enter a valid price!")
```

```python
add_product()
```

```python
def generate_report():

    print("\n===== GLAM FINDER PRODUCT REPORT =====")

    total_products = len(products)

    print("Total Products:", total_products)

    print("\nProducts by Category:")

    categories = []

    for product in products:
        if product["category"] not in categories:
            categories.append(product["category"])

    for category in categories:
        count = 0

        for product in products:
            if product["category"] == category:
                count = count + 1

        print(category, ":", count)

    print("\nProduct Report Generated Successfully!")
```

```python
generate_report()
```

```python
def user_login():

    username = input("Enter username: ")
    password = input("Enter password: ")

    if username == "glamuser" and password == "1234":
        print("\nLogin Successful!")
        return True

    else:
        print("\nInvalid username or password!")
        return False
```

```python
user_login()
```

```python
def main_menu():

    while True:

        print("\n================================")
        print("       GLAM FINDER")
        print("  Makeup Product Finder Portal")
        print("================================")

        print("1. View All Products")
        print("2. Search Product")
        print("3. Filter by Category")
        print("4. Filter by Budget")
        print("5. Filter by Skin Type")
        print("6. Add New Product")
        print("7. Generate Product Report")
        print("8. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            view_products()

        elif choice == "2":
            search_product()

        elif choice == "3":
            filter_category()

        elif choice == "4":
            filter_budget()

        elif choice == "5":
            filter_skin_type()

        elif choice == "6":
            add_product()

        elif choice == "7":
            generate_report()

        elif choice == "8":
            print("Thank you for using Glam Finder!")
            break

        else:
            print("Invalid choice! Please try again.")
```

```python
if user_login():
    main_menu()
else:
    print("Access Denied!")
```
